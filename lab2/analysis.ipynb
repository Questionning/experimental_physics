```python
import numpy as np
from scipy import stats

periods_10 = [
    [10.46, 10.45, 10.39, 10.94, 10.38],
    [12.44, 12.39, 12.38, 12.43, 12.28],
    [13.64, 13.78, 13.71, 13.72, 13.88],
    [15.70, 15.64, 15.41, 15.69, 15.65],
    [16.83, 16.78, 16.76, 16.88, 16.90],
    [14.45, 14.47, 14.51, 14.43, 14.52]
]

periods = np.array(periods_10) / 10
```

```python
table = []
for index, period in enumerate(periods):
    table.append([np.mean(period), np.mean(period), np.std(period, ddof=1)/np.sqrt(len(period))])

print(table)

```

```python
for i in table:
    print(f"Error on square mean = {2*i[0]*i[2]}")
```
