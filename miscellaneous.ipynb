```python
# Calculate the weighted mean

values = [3.03, 2.99, 2.99, 3.00, 3.05, 2.97]
uncertainties = [0.04, 0.03, 0.02, 0.05, 0.04, 0.02]
weights = [1/(u**2) for u in uncertainties]
mean = 0
for i in range(len(values)):
    mean += values[i] * weights[i]
mean /= sum(weights)
print(f"Expected mean: {mean} ± {math.sqrt(1/sum(weights))}")

# No there is a change by adding another point, but its weight is small 
# Calculate the weighted mean

values = [3.03, 2.99, 2.99, 3.00, 3.05, 2.97, 3.0]
uncertainties = [0.04, 0.03, 0.02, 0.05, 0.04, 0.02, 0.3]
weights = [1/(u**2) for u in uncertainties]
mean = 0
for i in range(len(values)):
    mean += values[i] * weights[i]
mean /= sum(weights)
print(f"Expected mean: {mean} ± {math.sqrt(1/sum(weights))}")

```
